```python
import pandas as pd
import matplotlib.pyplot as plt
import joblib
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv("Dengue_clinical_dataset.csv")
print(df.shape)
df.head()
```

```python
print(df.isna().sum().sum(), "missing values")
df["Outcome"].value_counts().plot.bar(color=["#e54", "#4c9"])
plt.title("Class distribution"); plt.xticks(rotation=0); plt.show()
```

```python
y = (df["Outcome"] == "Positive").astype(int)     # 1 = Positive, 0 = Negative
X = df.drop(columns=["Id", "Outcome"])            # Id is only a row number
X = pd.get_dummies(X, columns=["Gender", "Location"]).astype(float)
class_names = ["Negative", "Positive"]
X.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y)
print(X_train.shape, X_test.shape)
```

```python
tree = DecisionTreeClassifier(criterion="gini", max_depth=4, random_state=42)
tree.fit(X_train, y_train)
```

```python
y_pred = tree.predict(X_test)
print("Train accuracy:", round(accuracy_score(y_train, tree.predict(X_train)), 3))
print("Test accuracy: ", round(accuracy_score(y_test, y_pred), 3))
print(classification_report(y_test, y_pred, target_names=class_names))
```

```python
plt.figure(figsize=(20, 9))
plot_tree(tree, feature_names=X.columns, class_names=class_names, filled=True, rounded=True, fontsize=9)
plt.show()
```

```python
joblib.dump({
    "model": tree,
    "columns": list(X.columns),       # the exact columns the model expects
    "class_names": class_names,
}, "dengue_model.joblib")
print("Saved dengue_model.joblib")
```
